# Loading

```python
import pandas as pd 

data_citizen = pd.read_csv("B05001-Data.csv")
meta_citizen = pd.read_csv("B05001-Column-Metadata.csv")
```

```python
import pandas as pd 


data_foreign = pd.read_csv("B05006-Data.csv")
meta_foreign = pd.read_csv("B05006-Column-Metadata.csv")
```

# B05001 - Naturalized Citizens 

```python
print(data_citizen.shape)
print(data_citizen.head(3))
print(meta_citizen.shape)
```

```python
print(meta_citizen.to_string())
```

```python
df_citizen = data_citizen[['GEO_ID', 'B05001_005E', 'B05001_006E']].copy()

df_citizen = df_citizen.drop(index=0).reset_index(drop=True)
```

```python
df_citizen[['B05001_005E', 'B05001_006E']] = df_citizen[['B05001_005E', 'B05001_006E']].apply(pd.to_numeric, errors='coerce')

df_citizen = df_citizen.rename(columns={'B05001_005E': 'naturalized_citizens', 'B05001_006E': 'non_citizens'})
```

```python
df_citizen['GEO_ID'] = df_citizen['GEO_ID'].str.replace('1400000US', '')

print(df_citizen.shape)
print(df_citizen.head(5))
```

```python
print(df_citizen['naturalized_citizens'].sum())
print(df_citizen['non_citizens'].sum())
```

```python
df_citizen.to_csv('df_citizen_clean.csv', index=False)
```

# B05006 - Muslim Foreign Born  

```python
print(data_foreign.shape)
print(data_foreign.head(3))
print(meta_foreign.shape)
```

```python
data_foreign = data_foreign.drop(index=0).drop(columns=['Unnamed: 358']).reset_index(drop=True)
print(data_foreign.shape)
```

```python
print(meta_foreign.to_string())
```

```python
countries_to_find = [
    'Afghanistan', 'Bangladesh', 'Pakistan', 'Iran', 'Uzbekistan',
    'Indonesia', 'Azerbaijan', 'Iraq', 'Jordan', 'Kuwait',
    'Saudi Arabia', 'Syria', 'Turkey', 'Yemen', 'Somalia',
    'Algeria', 'Egypt', 'Morocco', 'Sudan', 'Senegal'
]

for country in countries_to_find:
    match = meta_foreign[meta_foreign['Label'].str.contains(country, case=False, na=False)]
    print(country, ':', match['Column Name'].values)
```

```python
muslim_cols = {
    'B05006_057E': 'Afghanistan',
    'B05006_058E': 'Bangladesh',
    'B05006_064E': 'Pakistan',
    'B05006_061E': 'Iran',
    'B05006_066E': 'Uzbekistan',
    'B05006_071E': 'Indonesia',
    'B05006_081E': 'Azerbaijan',
    'B05006_083E': 'Iraq',
    'B05006_085E': 'Jordan',
    'B05006_086E': 'Kuwait',
    'B05006_088E': 'Saudi Arabia',
    'B05006_089E': 'Syria',
    'B05006_090E': 'Turkey',
    'B05006_092E': 'Yemen',
    'B05006_100E': 'Somalia',
    'B05006_111E': 'Algeria',
    'B05006_112E': 'Egypt',
    'B05006_113E': 'Morocco',
    'B05006_114E': 'Sudan',
    'B05006_125E': 'Senegal',
}

df_muslim = data_foreign[['GEO_ID'] + list(muslim_cols.keys())].copy()
print(df_muslim.shape) 
```

```python
df_muslim = data_foreign[['GEO_ID'] + list(muslim_cols.keys())].copy()

df_muslim = df_muslim.drop(index=0).reset_index(drop=True)

df_muslim[list(muslim_cols.keys())] = df_muslim[list(muslim_cols.keys())].apply(pd.to_numeric, errors='coerce')

df_muslim['muslim_foreign_born'] = df_muslim[list(muslim_cols.keys())].sum(axis=1)
print(df_muslim.shape)
print(df_muslim[['GEO_ID', 'muslim_foreign_born']].head(40))
```

```python
print(df_muslim['muslim_foreign_born'].sum())
print(df_muslim['muslim_foreign_born'].max())
```

```python
total_foreign_born = pd.to_numeric(data_foreign.loc[1:, 'B05006_001E'], errors='coerce').sum()
print(total_foreign_born)
print(62338 / total_foreign_born * 100)
```

```python
df_muslim['GEO_ID'] = df_muslim['GEO_ID'].str.replace('1400000US', '')
print(df_muslim['GEO_ID'].head(5))
```

```python
df_muslim.to_csv('df_muslim_clean.csv', index=False)
```
