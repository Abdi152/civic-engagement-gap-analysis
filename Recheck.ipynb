# Double Checking Cook County Notebooks

```python
import pandas as pd

df_muslim = pd.read_csv("df_muslim_clean.csv")

# Check for duplicate GEOIDs
duplicate_geoids = df_muslim[df_muslim.duplicated(subset='GEO_ID', keep=False)]

print(f"Total rows: {len(df_muslim)}")
print(f"Unique GEOIDs: {df_muslim['GEO_ID'].nunique()}")
print(f"Duplicate rows found: {len(duplicate_geoids)}")

if len(duplicate_geoids) > 0:
    print("\nDuplicate GEOIDs:")
    print(duplicate_geoids)
```

```python
df_citizen = pd.read_csv("df_citizen_clean.csv")

duplicate_geoids_citizen = df_citizen[df_citizen.duplicated(subset='GEO_ID', keep=False)]

print(f"Total rows: {len(df_citizen)}")
print(f"Unique GEOIDs: {df_citizen['GEO_ID'].nunique()}")
print(f"Duplicate rows found: {len(duplicate_geoids_citizen)}")
```

```python
# Establishing ground truth totals to reconcile pipeline output against

county_muslim_total = df_muslim['muslim_foreign_born'].sum()
county_citizen_total = df_citizen['naturalized_citizens'].sum()

print(f"County-wide muslim_foreign_born total: {county_muslim_total}")
print(f"County-wide naturalized_citizens total: {county_citizen_total}")
```

```python
# Reconciliation check, confirms no population was lost across the two pipelines

ward_pop = pd.read_csv("ward_pop.csv")
township_pop = pd.read_csv("township_pop.csv")

ward_muslim_sum = ward_pop['muslim_foreign_born'].sum()
township_muslim_sum = township_pop['muslim_foreign_born'].sum()

combined_sum = ward_muslim_sum + township_muslim_sum
difference = county_muslim_total - combined_sum

print(f"Ward sum: {ward_muslim_sum}")
print(f"Township sum: {township_muslim_sum}")
print(f"Combined: {combined_sum}")
print(f"Difference from county total: {difference}")
```

```python
# Confirming the one remaining excluded tract (boundary-seam/water case) contributes 0 population, so excluding it doesn't bias totals

excluded_tract = df_muslim[df_muslim['GEO_ID'] == 17031836400]
print(excluded_tract[['GEO_ID', 'muslim_foreign_born']])
```

```python
ward_citizen_sum = ward_pop['naturalized_citizens'].sum()
township_citizen_sum = township_pop['naturalized_citizens'].sum()
combined_citizen = ward_citizen_sum + township_citizen_sum
print(f"Combined: {combined_citizen}, Difference from county total: {county_citizen_total - combined_citizen}")
```

# Double Checking DuPage County Notebook

```python
import pandas as pd

df_muslim_dupage = pd.read_csv("df_muslim_clean_dupage.csv")

# Check for duplicate GEOIDs
duplicate_geoids = df_muslim_dupage[df_muslim_dupage.duplicated(subset='GEO_ID', keep=False)]

print(f"Total rows: {len(df_muslim_dupage)}")
print(f"Unique GEOIDs: {df_muslim_dupage['GEO_ID'].nunique()}")
print(f"Duplicate rows found: {len(duplicate_geoids)}")

if len(duplicate_geoids) > 0:
    print("\nDuplicate GEOIDs:")
    print(duplicate_geoids)
```

```python
df_citizen_dupage = pd.read_csv("df_citizen_dupage_clean.csv")

duplicate_geoids_citizen = df_citizen_dupage[df_citizen_dupage.duplicated(subset='GEO_ID', keep=False)]

print(f"Total rows: {len(df_citizen_dupage)}")
print(f"Unique GEOIDs: {df_citizen_dupage['GEO_ID'].nunique()}")
print(f"Duplicate rows found: {len(duplicate_geoids_citizen)}")
```

```python
# Establishing ground truth totals to reconcile pipeline output against

county_muslim_total_dupage = df_muslim_dupage['muslim_foreign_born'].sum()
county_citizen_total_dupage = df_citizen_dupage['naturalized_citizens'].sum()

print(f"County-wide muslim_foreign_born total: {county_muslim_total_dupage}")
print(f"County-wide naturalized_citizens total: {county_citizen_total_dupage}")
```

```python

dupage_township_pop = pd.read_csv("dupage_township_pop_ranked.csv")

township_muslim_sum = dupage_township_pop['muslim_foreign_born'].sum()
difference = county_muslim_total_dupage - township_muslim_sum

print(f"Township sum: {township_muslim_sum}")
print(f"Difference from county total: {difference}")

township_citizen_sum = dupage_township_pop['naturalized_citizens'].sum()
difference_citizen = county_citizen_total_dupage - township_citizen_sum

print(f"Township citizen sum: {township_citizen_sum}")
print(f"Difference from county total: {difference_citizen}")
```
