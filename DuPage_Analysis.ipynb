# Loading

```python
import pandas as pd 

data_citizen_dupage = pd.read_csv("DuPage_B05001-Data.csv")
meta_citizen_dupage = pd.read_csv("DuPage_B05001-Column-Metadata.csv")
```

```python
import pandas as pd 

data_foreign_dupage = pd.read_csv("DuPage_B05006-Data.csv")
meta_foreign_dupage = pd.read_csv("DuPage_B05006-Column-Metadata.csv")
```

# B05001 - Natrualized Citizenship

```python
print(data_citizen_dupage.shape)
print(data_citizen_dupage.head(3))
print(meta_citizen_dupage.shape)
```

```python
print(meta_citizen_dupage.to_string())
```

```python
df_citizen_dupage = data_citizen_dupage[['GEO_ID', 'B05001_005E', 'B05001_006E']].copy()

df_citizen_dupage = df_citizen_dupage.drop(index=0).reset_index(drop=True)
```

```python
df_citizen_dupage[['B05001_005E', 'B05001_006E']] = df_citizen_dupage[['B05001_005E', 'B05001_006E']].apply(pd.to_numeric, errors='coerce')

df_citizen_dupage = df_citizen_dupage.rename(columns={'B05001_005E': 'naturalized_citizens', 'B05001_006E': 'non_citizens'})
```

```python
df_citizen_dupage['GEO_ID'] = df_citizen_dupage['GEO_ID'].str.replace('1400000US', '')

print(df_citizen_dupage.shape)
print(df_citizen_dupage.head(5))
```

```python
print(df_citizen_dupage['naturalized_citizens'].sum())
print(df_citizen_dupage['non_citizens'].sum())
```

```python
df_citizen_dupage.to_csv('df_citizen_dupage_clean.csv', index=False)
```

# B05006 - Muslim Foreign Born  

```python
print(data_foreign_dupage.shape)
print(data_foreign_dupage.head(3))
print(meta_foreign_dupage.shape)
```

```python
data_foreign_dupage = data_foreign_dupage.drop(index=0).drop(columns=['Unnamed: 358']).reset_index(drop=True)
print(data_foreign_dupage.shape)
```

```python
print(meta_foreign_dupage.to_string())
```

```python
countries_to_find = [
    'Afghanistan', 'Bangladesh', 'Pakistan', 'Iran', 'Uzbekistan',
    'Indonesia', 'Azerbaijan', 'Iraq', 'Jordan', 'Kuwait',
    'Saudi Arabia', 'Syria', 'Turkey', 'Yemen', 'Somalia',
    'Algeria', 'Egypt', 'Morocco', 'Sudan', 'Senegal'
]

for country in countries_to_find:
    match = meta_foreign_dupage[meta_foreign_dupage['Label'].str.contains(country, case=False, na=False)]
    print(country, ':', match['Column Name'].values)
```

```python
muslim_cols = {
    'B05006_057E': 'Afghanistan',
    'B05006_058E': 'Bangladesh',
    'B05006_064E': 'Pakistan',
    'B05006_061E': 'Iran',
    'B05006_066E': 'Uzbekistan',
    'B05006_071E': 'Indonesia',
    'B05006_081E': 'Azerbaijan',
    'B05006_083E': 'Iraq',
    'B05006_085E': 'Jordan',
    'B05006_086E': 'Kuwait',
    'B05006_088E': 'Saudi Arabia',
    'B05006_089E': 'Syria',
    'B05006_090E': 'Turkey',
    'B05006_092E': 'Yemen',
    'B05006_100E': 'Somalia',
    'B05006_111E': 'Algeria',
    'B05006_112E': 'Egypt',
    'B05006_113E': 'Morocco',
    'B05006_114E': 'Sudan',
    'B05006_125E': 'Senegal',
}

df_muslim_dupage = data_foreign_dupage[['GEO_ID'] + list(muslim_cols.keys())].copy()
print(df_muslim_dupage.shape) 
```

```python
df_muslim_dupage = data_foreign_dupage[['GEO_ID'] + list(muslim_cols.keys())].copy()

df_muslim_dupage[list(muslim_cols.keys())] = df_muslim_dupage[list(muslim_cols.keys())].apply(pd.to_numeric, errors='coerce')

df_muslim_dupage['muslim_foreign_born'] = df_muslim_dupage[list(muslim_cols.keys())].sum(axis=1)
print(df_muslim_dupage.shape)
print(df_muslim_dupage[['GEO_ID', 'muslim_foreign_born']].head(40))
```

```python
print(df_muslim_dupage['muslim_foreign_born'].sum())
print(df_muslim_dupage['muslim_foreign_born'].max())
```

```python
print(data_foreign_dupage[['GEO_ID','NAME']].head(2))
```

```python
total_foreign_born_dupage = pd.to_numeric(data_foreign_dupage['B05006_001E'], errors='coerce').sum()
print(total_foreign_born_dupage)
print(df_muslim_dupage['muslim_foreign_born'].sum() / total_foreign_born_dupage * 100)
```

```python
df_muslim_dupage['GEO_ID'] = df_muslim_dupage['GEO_ID'].str.replace('1400000US', '')
print(df_muslim_dupage['GEO_ID'].head(5))
```

```python
df_muslim_dupage.to_csv('df_muslim_clean_dupage.csv', index=False)
```
