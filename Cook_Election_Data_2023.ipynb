# Loading

```python
import pandas as pd 

election_data = pd.read_csv("election-data.csv")
```

# Election Data

```python
# Group precinct-level election data by ward, summing raw counts
ward_election = election_data.groupby('ward')[['registered', 'total']].sum().reset_index()

# Recalculate turnout on the aggregated (summed) numbers — not averaged from precinct percentages
ward_election['turnout'] = (ward_election['total'] / ward_election['registered']) * 100

# Quick sanity checks
print(ward_election.shape)   
print(ward_election.head())
print(ward_election['turnout'].describe())  
```

```python
ward_election.to_csv('ward_election_clean.csv', index=False)
print('saved:', ward_election.shape)
```
